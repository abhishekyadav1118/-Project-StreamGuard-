### Step 8: Detect Outliers Using the Interquartile Range (IQR) Method

**Purpose:** Detect statistical anomalies in numerical features using the 1.5 × IQR standard rule.

**Easy explanation:** The Interquartile Range (IQR) is the middle 50% of the data: $IQR = Q3 - Q1$. Any value below $Q1 - 1.5 \times IQR$ or above $Q3 + 1.5 \times IQR$ is flagged as an outlier.

**Line-by-line explanation:**
- `q1 = series.quantile(0.25)`: Calculates the 25th percentile (first quartile).
- `q3 = series.quantile(0.75)`: Calculates the 75th percentile (third quartile).
- `iqr = q3 - q1`: Computes the spread of the middle 50% of data.
- `lower_bound, upper_bound`: Defines the acceptable fence boundaries.
- `((series < lower_bound) | (series > upper_bound)).sum()`: Counts values outside the fences.

**Expected output:** A table showing Q1, Q3, IQR, boundary limits, and the number of outlier records detected per numerical feature.

**Presentation sentence:** *"Using the Interquartile Range method, we establish upper and lower statistical fences to detect anomalies across subscriber metrics."*


```python
# Step 8: IQR outlier detection
NUMERICAL_AUDIT_COLS = [
    "Watch_Hours", "Login_Frequency", "Support_Interactions",
    "Monthly_Fee", "Tenure_Months", "Customer_Age"
]

iqr_summary = []
for col in NUMERICAL_AUDIT_COLS:
    s = raw_df[col].dropna()
    q1 = s.quantile(0.25)
    q3 = s.quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    outlier_count = int(((s < lower) | (s > upper)).sum())
    iqr_summary.append({
        "Feature": col,
        "Q1": round(q1, 2),
        "Q3": round(q3, 2),
        "IQR": round(iqr, 2),
        "Lower Fence": round(lower, 2),
        "Upper Fence": round(upper, 2),
        "Outlier Count": outlier_count,
        "Outlier (%)": round(outlier_count / len(s) * 100, 2)
    })

iqr_df = pd.DataFrame(iqr_summary)
print("IQR Outlier Audit Report:")
display(iqr_df)
```

### Step 13: Create Domain-Specific Feature Engineering Transformer

**Purpose:** Implement a custom scikit-learn transformer to engineer subscriber engagement metrics.

**Easy explanation:** We engineer two domain-specific features:
1. **Engagement Index (`Engagement_Metric`)**: Combines viewing hours with login regularity: $\text{Watch\_Hours} \times (\text{Login\_Frequency} / 30)$.
2. **Engagement Efficiency Index (`Watch_to_Fee_Ratio`)**: Viewing hours per dollar spent: $\text{Watch\_Hours} / \max(\text{Monthly\_Fee}, 0.01)$.

**Line-by-line explanation:**
- `class OTTFeatureEngineer(BaseEstimator, TransformerMixin)`: Inherits from scikit-learn base classes so it plugs directly into scikit-learn Pipelines.
- `fit(self, X, y=None)`: Returns `self` because feature engineering does not need parameter learning.
- `transform(self, X)`: Computes the new ratio columns and appends them to the DataFrame.

**Expected output:** Confirmation that the custom transformer class is defined and ready.

**Presentation sentence:** *"We create a custom scikit-learn transformer to engineer two domain metrics: an overall Engagement Index and a Watch-to-Fee efficiency ratio."*


```python
# Step 13: Define feature engineering transformer
from sklearn.base import BaseEstimator, TransformerMixin

try:
    from streamguard_transformers import OTTFeatureEngineer
except ImportError:
    class OTTFeatureEngineer(BaseEstimator, TransformerMixin):
        """Generates Engagement_Metric and Watch_to_Fee_Ratio."""
        def fit(self, X, y=None):
            return self  # No parameters to learn during fit
        
        def transform(self, X):
            X = pd.DataFrame(X).copy()
            watch = pd.to_numeric(X["Watch_Hours"], errors="coerce").fillna(0)
            login = pd.to_numeric(X["Login_Frequency"], errors="coerce").fillna(0)
            fee = pd.to_numeric(X["Monthly_Fee"], errors="coerce").fillna(10)
            
            # 1. Engagement Index: total viewing scaled by monthly login frequency
            X["Engagement_Metric"] = np.round(watch * (login / 30.0), 3)
            
            # 2. Engagement Efficiency: hours watched per dollar spent
            X["Watch_to_Fee_Ratio"] = np.round(watch / np.maximum(fee, 0.01), 3)
            return X

# Add engineered column names to numerical features list
ENGINEERED_NUMERICAL_COLS = ["Engagement_Metric", "Watch_to_Fee_Ratio"]
ALL_NUMERICAL_COLS = BASE_NUMERICAL_COLS + ENGINEERED_NUMERICAL_COLS

print("Step 13 Complete: OTTFeatureEngineer defined.")
print(f"Total Numerical Features after Engineering ({len(ALL_NUMERICAL_COLS)}): {ALL_NUMERICAL_COLS}")
```

### Step 14: Define Custom IQR Outlier Clipping Transformer

**Purpose:** Implement a transformer that caps extreme outliers to training-set statistical fences.

**Easy explanation:** Rather than discarding rows, this transformer learns the upper and lower fences on the training data during `.fit()`, and clips extreme values during `.transform()`. This prevents extreme anomalies from distorting calculations.

**Line-by-line explanation:**
- `fit(self, X, y=None)`: Calculates and stores the IQR fence limits for each numerical column on the training partition.
- `transform(self, X)`: Applies `.clip(lower, upper)` using the stored bounds learned during `fit()`.

**Expected output:** Confirmation that the `IQRClipper` transformer is ready.

**Presentation sentence:** *"We define an `IQRClipper` transformer that learns statistical fences strictly during `fit()` on training data, preventing data leakage."*


```python
# Step 14: Define IQR outlier clipping transformer
try:
    from streamguard_transformers import IQRClipper
except ImportError:
    class IQRClipper(BaseEstimator, TransformerMixin):
        """Clips numerical values to IQR fences learned during fit()."""
        def __init__(self, columns=None, multiplier=1.5):
            self.columns = columns
            self.multiplier = multiplier
        
        def fit(self, X, y=None):
            X = pd.DataFrame(X)
            cols = self.columns or X.columns.tolist()
            self.bounds_ = {}
            for c in cols:
                s = pd.to_numeric(X[c], errors="coerce").dropna()
                if len(s) == 0:
                    continue
                q1 = s.quantile(0.25)
                q3 = s.quantile(0.75)
                iqr = q3 - q1
                self.bounds_[c] = (q1 - self.multiplier * iqr, q3 + self.multiplier * iqr)
            return self
        
        def transform(self, X):
            X = pd.DataFrame(X).copy()
            for c, (lower, upper) in self.bounds_.items():
                if c in X.columns:
                    X[c] = pd.to_numeric(X[c], errors="coerce").clip(lower, upper)
            return X

print("Step 14 Complete: IQRClipper transformer defined successfully.")
```

### Step 15: Build the Complete Leak-Free Preprocessing Pipeline

**Purpose:** Combine median imputation, outlier clipping, Z-score scaling, and One-Hot Encoding into a single scikit-learn `Pipeline`.

**Easy explanation:** A `Pipeline` chains all transformation steps together into one object.
- Numerical branch: Imputes missing values with **median**, clips outliers with **IQRClipper**, and scales with **StandardScaler** ($z = (x - \mu) / \sigma$).
- Categorical branch: Imputes missing values with **mode** (most frequent), and converts categories into dummy 0/1 columns using **OneHotEncoder**.

**Line-by-line explanation:**
- `SimpleImputer(strategy="median")`: Fills numerical NaNs with the middle value.
- `StandardScaler()`: Standardizes features to mean = 0 and variance = 1.
- `OneHotEncoder(handle_unknown="ignore")`: Turns categories into binary indicator columns.
- `ColumnTransformer(...)`: Directs numerical columns to the numerical pipeline and categorical columns to the categorical pipeline.

**Expected output:** Confirmation that the complete preprocessor pipeline is built.

**Presentation sentence:** *"Using `ColumnTransformer`, we integrate median imputation, StandardScaler, and OneHotEncoder into a single pipeline to ensure zero data leakage."*


```python
# Step 15: Construct preprocessing pipeline
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

def build_preprocessor():
    """Builds complete leak-free preprocessing pipeline."""
    # Numerical sub-pipeline: Median Imputation -> IQR Clipping -> StandardScaler
    num_pipeline = Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("iqr_clipper", IQRClipper(multiplier=1.5)),
        ("scaler", StandardScaler()),
    ])
    
    # Categorical sub-pipeline: Mode Imputation -> One-Hot Encoding
    cat_pipeline = Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("encoder", OneHotEncoder(handle_unknown="ignore", sparse_output=False)),
    ])
    
    # ColumnTransformer routes each column group to its respective sub-pipeline
    column_transformer = ColumnTransformer([
        ("num", num_pipeline, ALL_NUMERICAL_COLS),
        ("cat", cat_pipeline, CATEGORICAL_COLS),
    ], remainder="drop")
    
    # Full preprocessor: Feature Engineering first, then ColumnTransformer
    full_preprocessor = Pipeline([
        ("feature_engineer", OTTFeatureEngineer()),
        ("columns", column_transformer),
    ])
    
    return full_preprocessor

preprocessor = build_preprocessor()
print("Step 15 Complete: Full preprocessing pipeline constructed.")
```

```python
# Step 17: Train / Test split with stratification
from sklearn.model_selection import train_test_split

X = df[FEATURE_COLS].copy()
y_churn = df[TARGET_CLASSIFICATION].copy()
y_clv = df[TARGET_REGRESSION].copy()

X_train, X_test, y_churn_train, y_churn_test, y_clv_train, y_clv_test = train_test_split(
    X, y_churn, y_clv, test_size=0.20, random_state=RANDOM_STATE, stratify=y_churn
)

print(f"Training Set Rows: {X_train.shape[0]} (Churn Rate: {y_churn_train.mean():.2%})")
print(f"Held-Out Test Set Rows: {X_test.shape[0]} (Churn Rate: {y_churn_test.mean():.2%})")
print("Note: The test set is locked and will NOT be touched until final evaluation in Module 5.")
```

### Step 18: Set Up Cross-Validation Schemes (Stratified K-Fold and K-Fold)

**Purpose:** Configure cross-validation splitters for both classification and regression.

**Easy explanation:** Cross-validation divides the training data into $K=5$ equal folds. In each round, 4 folds train the model and 1 fold tests it. This is repeated 5 times so every data point is tested once, giving us a reliable, unbiased performance score.

**Line-by-line explanation:**
- `StratifiedKFold(n_splits=5, shuffle=True, random_state=42)`: Used for classification so each of the 5 folds contains the same percentage of churners.
- `KFold(n_splits=5, shuffle=True, random_state=42)`: Used for continuous CLV regression where target values are continuous numbers.

**Expected output:** Confirmation showing that 5-fold cross-validation schemes are configured.

**Presentation sentence:** *"We establish 5-fold Stratified Cross-Validation for classification to maintain class proportions, and 5-fold K-Fold for our continuous CLV regression."*


```python
# Step 18: Set up cross-validation splitters
from sklearn.model_selection import StratifiedKFold, KFold

# 5-Fold Stratified CV for Churn Classification
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)

# 5-Fold Standard K-Fold for CLV Regression
kf = KFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)

print("Step 18 Complete: 5-Fold StratifiedKFold and KFold splitters initialized.")
```

### Step 19: Apply Recursive Feature Elimination (RFE) Inside Cross-Validation

**Purpose:** Use Recursive Feature Elimination (RFE) to identify the top 15 most important features and compare performance against using all 24 features.

**Easy explanation:** RFE starts with all features, fits an estimator, ranks feature importance, and discards the weakest features one by one until only the top 15 remain. Fitting RFE inside cross-validation guarantees zero data leakage.

**Line-by-line explanation:**
- `RFE(estimator=RandomForestClassifier(...), n_features_to_select=15)`: Iteratively prunes features to select the best 15.
- `pipe_all` vs `pipe_rfe`: Compares using all 24 features against the RFE-selected subset across 5 cross-validation folds.

**Expected output:** A comparison table showing Accuracy, F1-Score, and ROC-AUC for all features vs RFE-selected features.

**Presentation sentence:** *"We run Recursive Feature Elimination inside our cross-validation loop, comparing all 24 features against the top 15 RFE features to confirm predictive stability."*


```python
# Step 19: Recursive Feature Elimination (RFE) inside CV
from sklearn.feature_selection import RFE
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, f1_score, roc_auc_score
from sklearn.base import clone

base_lr = LogisticRegression(max_iter=1000, class_weight="balanced", random_state=RANDOM_STATE)

# Pipeline A: All 24 Features
pipe_all = Pipeline([("prep", build_preprocessor()), ("clf", clone(base_lr))])

# Pipeline B: Top 15 Features selected via RFE
pipe_rfe = Pipeline([
    ("prep", build_preprocessor()),
    ("rfe", RFE(RandomForestClassifier(n_estimators=100, random_state=RANDOM_STATE, n_jobs=-1),
                n_features_to_select=15, step=0.1)),
    ("clf", clone(base_lr)),
])

rfe_benchmark = []
for label, pipeline in [("All Features (24)", pipe_all), ("RFE Selected (15)", pipe_rfe)]:
    scores = []
    for tr_idx, va_idx in skf.split(X_train, y_churn_train):
        m = clone(pipeline).fit(X_train.iloc[tr_idx], y_churn_train.iloc[tr_idx])
        preds = m.predict(X_train.iloc[va_idx])
        probs = m.predict_proba(X_train.iloc[va_idx])[:, 1]
        y_val = y_churn_train.iloc[va_idx]
        scores.append({
            "acc": accuracy_score(y_val, preds),
            "f1": f1_score(y_val, preds, zero_division=0),
            "roc": roc_auc_score(y_val, probs),
        })
    df_s = pd.DataFrame(scores)
    rfe_benchmark.append({
        "Feature Set": label,
        "Accuracy (Mean)": round(df_s["acc"].mean(), 4),
        "F1-Score (Mean)": round(df_s["f1"].mean(), 4),
        "ROC-AUC (Mean)": round(df_s["roc"].mean(), 4),
    })

print("RFE Feature Selection Benchmark Results:")
display(pd.DataFrame(rfe_benchmark))
```

### Step 23: Evaluate Baseline Classifiers on 5-Fold Cross-Validation

**Purpose:** Compare the performance of the three baselines across 5 cross-validation folds on the training set.

**Easy explanation:** We compare Logistic Regression, KNN, and SVM on Accuracy, F1-Score, and ROC-AUC. F1-Score and ROC-AUC are our primary metrics because accuracy alone is misleading on imbalanced datasets.

**Line-by-line explanation:**
- `cross_val_score` / manual fold loop: Evaluates each pipeline on unseen validation folds.
- `roc_auc_score`: Measures how well the model separates churners from non-churners across all possible thresholds.

**Expected output:** A summary table showing cross-validation metrics for all three baseline models.

**Presentation sentence:** *"We benchmark our three baselines using 5-fold cross-validation, observing that Logistic Regression achieves the highest baseline ROC-AUC of 0.777."*


```python
# Step 23: Compare baseline classifiers via 5-fold CV
from sklearn.metrics import precision_score, recall_score

baseline_dict = {
    "Logistic Regression": pipe_lr,
    "K-Nearest Neighbors (KNN)": pipe_knn,
    "Support Vector Machine (SVM)": pipe_svm,
}

baseline_cv_results = []
for name, pipe in baseline_dict.items():
    f_scores = []
    for tr, va in skf.split(X_train, y_churn_train):
        m = clone(pipe).fit(X_train.iloc[tr], y_churn_train.iloc[tr])
        preds = m.predict(X_train.iloc[va])
        probs = m.predict_proba(X_train.iloc[va])[:, 1]
        yv = y_churn_train.iloc[va]
        f_scores.append({
            "acc": accuracy_score(yv, preds),
            "prec": precision_score(yv, preds, zero_division=0),
            "rec": recall_score(yv, preds, zero_division=0),
            "f1": f1_score(yv, preds, zero_division=0),
            "roc": roc_auc_score(yv, probs),
        })
    df_f = pd.DataFrame(f_scores)
    baseline_cv_results.append({
        "Baseline Model": name,
        "Accuracy (Mean)": round(df_f["acc"].mean(), 4),
        "Precision (Mean)": round(df_f["prec"].mean(), 4),
        "Recall (Mean)": round(df_f["rec"].mean(), 4),
        "F1-Score (Mean)": round(df_f["f1"].mean(), 4),
        "ROC-AUC (Mean)": round(df_f["roc"].mean(), 4),
    })

print("Baseline Classification Comparison (5-Fold CV on Training Set):")
display(pd.DataFrame(baseline_cv_results))
```

### Step 24: Configure Hyperparameter Tuning via RandomizedSearchCV

**Purpose:** Set up `RandomizedSearchCV` to optimize hyperparameters for Random Forest and AdaBoost.

**Easy explanation:** Hyperparameters are settings we choose before training (like tree depth or number of estimators). Instead of testing every combination which takes too long, `RandomizedSearchCV` samples a diverse set of candidate combinations across 3 internal folds, efficiently finding optimal configurations.

**Line-by-line explanation:**
- `param_distributions`: Dictionary defining candidate values for `n_estimators`, `max_depth`, `min_samples_split`, and `learning_rate`.
- `RandomizedSearchCV(n_iter=6, scoring="roc_auc", cv=3)`: Tries 6 random parameter combinations, evaluating each on 3 inner cross-validation folds.

**Expected output:** Confirmation that search spaces are configured for Random Forest and AdaBoost.

**Presentation sentence:** *"We configure RandomizedSearchCV to tune tree depth, estimator counts, and learning rates across inner cross-validation folds."*


```python
# Step 24: Configure RandomizedSearchCV for ensemble models
from sklearn.model_selection import RandomizedSearchCV
from sklearn.ensemble import AdaBoostClassifier

inner_cv = StratifiedKFold(n_splits=3, shuffle=True, random_state=RANDOM_STATE)

# 1. Random Forest Parameter Search Space
rf_pipeline = Pipeline([
    ("prep", build_preprocessor()),
    ("clf", RandomForestClassifier(class_weight="balanced", random_state=RANDOM_STATE, n_jobs=-1))
])
rf_param_grid = {
    "clf__n_estimators": [100, 150, 200],
    "clf__max_depth": [None, 8, 12, 16],
    "clf__min_samples_split": [2, 4, 8],
    "clf__min_samples_leaf": [1, 2, 4],
}
rf_search = RandomizedSearchCV(
    rf_pipeline, rf_param_grid, n_iter=6, scoring="roc_auc",
    cv=inner_cv, random_state=RANDOM_STATE, n_jobs=-1, refit=True
)

# 2. AdaBoost Parameter Search Space
ab_pipeline = Pipeline([
    ("prep", build_preprocessor()),
    ("clf", AdaBoostClassifier(random_state=RANDOM_STATE))
])
ab_param_grid = {
    "clf__n_estimators": [50, 100, 150, 200],
    "clf__learning_rate": [0.01, 0.05, 0.1, 0.5, 1.0],
}
ab_search = RandomizedSearchCV(
    ab_pipeline, ab_param_grid, n_iter=6, scoring="roc_auc",
    cv=inner_cv, random_state=RANDOM_STATE, n_jobs=-1, refit=True
)

print("Step 24 Complete: RandomizedSearchCV configured for Random Forest and AdaBoost.")
```

### Step 25: Train Baseline Regression Models for Customer Lifetime Value (CLV)

**Purpose:** Train and evaluate Linear Regression versus Random Forest Regressor on the continuous target `Lifetime_Value_USD`.

**Easy explanation:** To predict the total dollar amount a customer will generate, we benchmark two regression models using 5-fold cross-validation. We evaluate with MAE (Mean Absolute Error), RMSE (Root Mean Squared Error), and $R^2$ (coefficient of determination).

**Line-by-line explanation:**
- `LinearRegression()`: Standard ordinary least squares regression model.
- `RandomForestRegressor(n_estimators=150, max_depth=12)`: Non-linear ensemble of 150 regression trees.
- `mean_squared_error(..., squared=False)` / `np.sqrt(...)`: Computes RMSE in US Dollars.

**Expected output:** Cross-validation comparison table for the two regression models.

**Presentation sentence:** *"For continuous Customer Lifetime Value prediction, we evaluate Linear Regression against Random Forest Regressor using 5-fold CV, measuring MAE, RMSE, and R²."*


```python
# Step 25: CLV regression cross-validation benchmark
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

regression_models = {
    "Linear Regression": Pipeline([("prep", build_preprocessor()), ("reg", LinearRegression())]),
    "Random Forest Regressor": Pipeline([
        ("prep", build_preprocessor()),
        ("reg", RandomForestRegressor(n_estimators=150, max_depth=12, random_state=RANDOM_STATE, n_jobs=-1))
    ])
}

reg_cv_summary = []
for name, pipe in regression_models.items():
    fold_evals = []
    for tr, va in kf.split(X_train):
        m = clone(pipe).fit(X_train.iloc[tr], y_clv_train.iloc[tr])
        preds = m.predict(X_train.iloc[va])
        y_val_clv = y_clv_train.iloc[va]
        fold_evals.append({
            "MAE": mean_absolute_error(y_val_clv, preds),
            "RMSE": np.sqrt(mean_squared_error(y_val_clv, preds)),
            "R2": r2_score(y_val_clv, preds),
        })
    df_evals = pd.DataFrame(fold_evals)
    reg_cv_summary.append({
        "Model": name,
        "MAE (Mean $)": round(df_evals["MAE"].mean(), 2),
        "RMSE (Mean $)": round(df_evals["RMSE"].mean(), 2),
        "R2 (Mean)": round(df_evals["R2"].mean(), 4),
    })

print("CLV Regression 5-Fold Cross-Validation Benchmark:")
display(pd.DataFrame(reg_cv_summary))
```

# MODULE 5: ENSEMBLE LEARNING AND FINAL EVALUATION

In this final module, we compare all 5 classification models using leakage-safe nested cross-validation, optimize the decision threshold on training data, evaluate on the held-out test set, and provide business recommendations.


### Step 26: Execute Leakage-Safe Nested Cross-Validation Across All 5 Classifiers

**Purpose:** Compare all 5 models (Logistic Regression, KNN, SVM, Random Forest, AdaBoost) using nested 5-fold cross-validation.

**Easy explanation:** In **nested cross-validation**, parameter tuning is nested strictly inside each outer training fold. The outer validation fold is completely unseen by both the transformer and the search grid. This guarantees an unbiased comparison.

**Line-by-line explanation:**
- Outer loop (`outer_skf.split(X_train)`): 5 outer splits for unbiased evaluation.
- Inner loop (`RandomizedSearchCV`): Tunes hyperparameters exclusively on `X_train.iloc[tr]`.
- Out-of-fold probability predictions (`oof_predictions`): Collected for training threshold tuning.

**Expected output:** Master comparison table showing 5-fold CV metrics (Accuracy, Precision, Recall, F1, ROC-AUC) across all 5 models.

**Presentation sentence:** *"We run leakage-safe nested cross-validation across all five models, ensuring hyperparameter tuning occurs strictly within training folds."*


```python
# Step 26: Leakage-safe nested cross-validation
all_classifiers = {
    "Logistic Regression": baseline_dict["Logistic Regression"],
    "K-Nearest Neighbors (KNN)": baseline_dict["K-Nearest Neighbors (KNN)"],
    "Support Vector Machine (SVM)": baseline_dict["Support Vector Machine (SVM)"],
    "Random Forest (Tuned Nested CV)": rf_search,
    "AdaBoost (Tuned Nested CV)": ab_search,
}

nested_cv_table = []
oof_probs_dict = {}

print("Executing 5-Fold Nested Cross-Validation (this runs in ~30 seconds)...")
for name, model_pipeline in all_classifiers.items():
    fold_scores = []
    oof_probs = np.zeros(len(X_train))
    
    for tr_idx, va_idx in skf.split(X_train, y_churn_train):
        X_tr, y_tr = X_train.iloc[tr_idx], y_churn_train.iloc[tr_idx]
        X_va, y_va = X_train.iloc[va_idx], y_churn_train.iloc[va_idx]
        
        # Fit model (inner CV search occurs strictly on X_tr)
        m = clone(model_pipeline).fit(X_tr, y_tr)
        preds = m.predict(X_va)
        probs = m.predict_proba(X_va)[:, 1] if hasattr(m, "predict_proba") else preds
        oof_probs[va_idx] = probs
        
        fold_scores.append({
            "acc": accuracy_score(y_va, preds),
            "prec": precision_score(y_va, preds, zero_division=0),
            "rec": recall_score(y_va, preds, zero_division=0),
            "f1": f1_score(y_va, preds, zero_division=0),
            "roc": roc_auc_score(y_va, probs),
        })
    
    oof_probs_dict[name] = oof_probs
    df_s = pd.DataFrame(fold_scores)
    nested_cv_table.append({
        "Model": name,
        "Accuracy (Mean)": round(df_s["acc"].mean(), 4),
        "Precision (Mean)": round(df_s["prec"].mean(), 4),
        "Recall (Mean)": round(df_s["rec"].mean(), 4),
        "F1-Score (Mean)": round(df_s["f1"].mean(), 4),
        "ROC-AUC (Mean)": round(df_s["roc"].mean(), 4),
        "ROC-AUC (Std)": round(df_s["roc"].std(), 4),
    })

comparison_df = pd.DataFrame(nested_cv_table).sort_values("ROC-AUC (Mean)", ascending=False).reset_index(drop=True)
print("\n=== 5-MODEL CROSS-VALIDATION BENCHMARK RESULTS ===")
display(comparison_df)

best_model_name = comparison_df.iloc[0]["Model"]
print(f"\nSelected Winning Model: {best_model_name} (Highest CV ROC-AUC: {comparison_df.iloc[0]['ROC-AUC (Mean)']})")
```

### Step 27: Select Decision Threshold on Training Cross-Validation Only

**Purpose:** Optimize the classification decision threshold strictly on training out-of-fold predictions to balance precision and recall.

**Easy explanation:** By default, classifiers use a threshold of 0.50 (probabilities $\ge 0.50$ are flagged as churn). But in subscription retention, catching more churners (recall) is crucial. We evaluate candidate thresholds from 0.10 to 0.85 on training CV to find the threshold that maximizes F1-Score. The test set remains untouched!

**Line-by-line explanation:**
- `threshold_candidates = np.arange(0.10, 0.85, 0.05)`: Tests thresholds at 0.05 step increments.
- `(best_oof_probs >= t).astype(int)`: Converts probabilities into 0/1 predictions at each threshold.
- `threshold_results.sort_values("F1-Score", ascending=False)`: Finds the threshold giving the optimal F1-score.

**Expected output:** Table of thresholds with their corresponding Precision, Recall, and F1-Scores, and the chosen optimal threshold (0.60).

**Presentation sentence:** *"Using training out-of-fold predictions, we analyze the precision-recall trade-off and select an optimal decision threshold of 0.60 without touching the test set."*


```python
# Step 27: Precision-Recall trade-off & threshold optimization on training CV
best_oof_probs = oof_probs_dict[best_model_name]

threshold_candidates = np.arange(0.10, 0.85, 0.05)
threshold_records = []

for t in threshold_candidates:
    binary_preds = (best_oof_probs >= t).astype(int)
    threshold_records.append({
        "Threshold": round(t, 2),
        "Precision": round(precision_score(y_churn_train, binary_preds, zero_division=0), 4),
        "Recall": round(recall_score(y_churn_train, binary_preds, zero_division=0), 4),
        "F1-Score": round(f1_score(y_churn_train, binary_preds, zero_division=0), 4),
    })

threshold_df = pd.DataFrame(threshold_records)
optimal_threshold_row = threshold_df.sort_values(["F1-Score", "Recall"], ascending=False).iloc[0]
OPTIMAL_THRESHOLD = float(optimal_threshold_row["Threshold"])

print("Training CV Threshold Optimization Table:")
display(threshold_df.head(10))
print(f"\nSelected Optimal Decision Threshold: {OPTIMAL_THRESHOLD:.2f}")
print(f"Training CV Metrics at Threshold {OPTIMAL_THRESHOLD:.2f}: "
      f"Precision = {optimal_threshold_row['Precision']}, "
      f"Recall = {optimal_threshold_row['Recall']}, "
      f"F1-Score = {optimal_threshold_row['F1-Score']}")
```

### Step 28: Fit Final Model on Full Training Set and Evaluate on Held-Out Test Set

**Purpose:** Fit the selected model pipeline on the entire training set (`X_train`), and evaluate it once on the held-out test set (`X_test`).

**Easy explanation:** Now that all design decisions (features, model type, threshold) are finalized, we train our winning pipeline on 100% of the training data and run our final test on the 301 held-out test customers.

**Line-by-line explanation:**
- `final_pipeline.fit(X_train, y_churn_train)`: Fits the pipeline on all training records.
- `test_probs = final_pipeline.predict_proba(X_test)[:, 1]`: Generates predicted churn probabilities on test data.
- Evaluates at both default 0.50 threshold and optimal 0.60 threshold.

**Expected output:** A table comparing test metrics at threshold 0.50 vs threshold 0.60.

**Presentation sentence:** *"We train our final pipeline on the complete training set and evaluate on the untouched held-out test set, confirming strong generalization with 0.845 ROC-AUC."*


```python
# Step 28: Final model test set evaluation
final_pipeline = clone(all_classifiers[best_model_name])
final_pipeline.fit(X_train, y_churn_train)

# Predict on held-out test set
test_probs = final_pipeline.predict_proba(X_test)[:, 1]
test_preds_default = (test_probs >= 0.50).astype(int)
test_preds_optimal = (test_probs >= OPTIMAL_THRESHOLD).astype(int)

test_metrics_comparison = pd.DataFrame([
    {
        "Evaluation": "Standard Threshold (0.50)",
        "Accuracy": round(accuracy_score(y_churn_test, test_preds_default), 4),
        "Precision": round(precision_score(y_churn_test, test_preds_default, zero_division=0), 4),
        "Recall": round(recall_score(y_churn_test, test_preds_default, zero_division=0), 4),
        "F1-Score": round(f1_score(y_churn_test, test_preds_default, zero_division=0), 4),
        "ROC-AUC": round(roc_auc_score(y_churn_test, test_probs), 4),
    },
    {
        "Evaluation": f"Optimal Threshold ({OPTIMAL_THRESHOLD:.2f})",
        "Accuracy": round(accuracy_score(y_churn_test, test_preds_optimal), 4),
        "Precision": round(precision_score(y_churn_test, test_preds_optimal, zero_division=0), 4),
        "Recall": round(recall_score(y_churn_test, test_preds_optimal, zero_division=0), 4),
        "F1-Score": round(f1_score(y_churn_test, test_preds_optimal, zero_division=0), 4),
        "ROC-AUC": round(roc_auc_score(y_churn_test, test_probs), 4),
    }
])

print(f"Final Held-Out Test Evaluation ({best_model_name}):")
display(test_metrics_comparison)
```
